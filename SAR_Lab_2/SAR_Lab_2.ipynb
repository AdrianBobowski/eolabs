**QUESTION (10pts): What is the line of sight angle at the near range of the scene and at the far range of the scene? Which edge of the scene (east/west) will be more susceptible to foreshortening/layover effects, and which edge of the scene will be more susceptible to shadow effects?**

**ANSWER:** 
The line of sight (incidence) angle varies across a Sentinel-1 scene from the near range (closest to the satellite track, where the angle is smaller, typically around $20^\circ\text{--}30^\circ$) to the far range (farthest from the satellite track, where the angle is larger, typically around $40^\circ\text{--}45^\circ$). Depending on the satellite's orbital pass, the susceptibility to radar distortions changes across the scene. In an ascending orbit, where the satellite moves north and looks to the east, the west edge represents the near range with a smaller incidence angle, making it more susceptible to foreshortening and layover effects, while the east edge represents the far range with a larger incidence angle, making it more susceptible to shadow effects. Conversely, in a descending orbit, where the satellite moves south and looks to the west, the east edge becomes the near range with a smaller incidence angle, increasing its susceptibility to foreshortening and layover effects, whereas the west edge becomes the far range with a larger incidence angle, making it more susceptible to shadow effects.

## 2. Helper functions

These functions keep the notebook readable and reduce repeated plotting code.


```python
def read_raster(path, band=1):
    """Read a single-band raster and return array + metadata.
    Nodata values are converted to NaN.
    """
    with rasterio.open(path) as src:
        arr = src.read(band).astype("float32")
        meta = src.meta.copy()
        bounds = src.bounds
        transform = src.transform
        crs = src.crs
        nodata = src.nodata

    if nodata is not None:
        arr[arr == nodata] = np.nan

    return arr, meta, bounds, transform, crs


def plot_image(arr, title="", vmin=None, vmax=None, cmap="gray", figsize=(8, 6)):
    plt.figure(figsize=figsize)
    plt.imshow(arr, cmap=cmap, vmin=vmin, vmax=vmax)
    plt.colorbar()
    plt.title(title)
    plt.axis("off")
    plt.show()


def plot_side_by_side(arr1, arr2, title1, title2, vmin=None, vmax=None, cmap="gray", figsize=(14, 6)):
    fig, axes = plt.subplots(1, 2, figsize=figsize)
    axes[0].imshow(arr1, cmap=cmap, vmin=vmin, vmax=vmax)
    axes[0].set_title(title1)
    axes[0].axis("off")

    im2 = axes[1].imshow(arr2, cmap=cmap, vmin=vmin, vmax=vmax)
    axes[1].set_title(title2)
    axes[1].axis("off")

    fig.colorbar(im2, ax=axes.ravel().tolist(), shrink=0.8)
    plt.show()


def to_db(arr, eps=1e-6):
    """Convert linear power to dB if needed.
    If your exported rasters are already in dB, do not apply this.
    """
    return 10 * np.log10(np.maximum(arr, eps))


def robust_limits(arr, pmin=2, pmax=98):
    valid = arr[np.isfinite(arr)]
    return np.percentile(valid, pmin), np.percentile(valid, pmax)


def histogram(arr, bins=100, title="", xlabel="", logy=False):
    valid = arr[np.isfinite(arr)]
    plt.figure(figsize=(8, 4))
    plt.hist(valid, bins=bins)
    if logy:
        plt.yscale("log")
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel("Pixel count")
    plt.grid(True, alpha=0.3)
    plt.xlim(xmin=0,xmax=200)
    plt.show()

def plot_transects(arr1, arr2, label1="Before TC", label2="After TC", axis=0, title="Central transect"):
    t1 = central_transect(arr1, axis=axis)
    t2 = central_transect(arr2, axis=axis)

    plt.figure(figsize=(10, 4))
    plt.plot(t1, label=label1)
    plt.plot(t2, label=label2)
    plt.title(title)
    plt.xlabel("Pixel index")
    plt.ylabel("Backscatter")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.show()

def pick_points(image, title="Click 2 points, then close the figure", cmap="gray", vmin=None, vmax=None):
    """Display an image and let the user click two points.

    Returns:
        pts: list of two (x, y) integer pixel tuples
    """
    fig, ax = plt.subplots(figsize=(10, 8))
    ax.imshow(image, cmap=cmap, vmin=vmin, vmax=vmax)
    ax.set_title(title)
    ax.set_xlabel("X pixel")
    ax.set_ylabel("Y pixel")

    pts = plt.ginput(2, timeout=-1)
    plt.close(fig)

    pts = [(int(round(x)), int(round(y))) for x, y in pts]
    return pts


def extract_transect(arr, p0, p1, n=500):
    """Extract values along a line between two pixel coordinates."""
    x0, y0 = p0
    x1, y1 = p1

    xs = np.linspace(x0, x1, n)
    ys = np.linspace(y0, y1, n)

    xs_i = np.clip(np.round(xs).astype(int), 0, arr.shape[1] - 1)
    ys_i = np.clip(np.round(ys).astype(int), 0, arr.shape[0] - 1)

    values = arr[ys_i, xs_i]
    distance = np.arange(n)

    return distance, values, xs_i, ys_i


def plot_transect_on_image(image, p0, p1, title="", cmap="gray", vmin=None, vmax=None):
    """Plot an image with the selected transect overlaid."""
    fig, ax = plt.subplots(figsize=(10, 8))
    ax.imshow(image, cmap=cmap, vmin=vmin, vmax=vmax)
    ax.plot([p0[0], p1[0]], [p0[1], p1[1]], 'r-', linewidth=2)
    ax.plot(p0[0], p0[1], 'go', markersize=8, label='Start')
    ax.plot(p1[0], p1[1], 'bo', markersize=8, label='End')
    ax.set_title(title)
    ax.set_xlabel("X pixel")
    ax.set_ylabel("Y pixel")
    ax.legend()
    plt.show()


def robust_limits(arr, pmin=2, pmax=98):
    valid = arr[np.isfinite(arr)]
    return np.percentile(valid, pmin), np.percentile(valid, pmax)
```

**QUESTION (10pts): Why is the CRS (coordinate reference system) for the Pre-TC view "None" while the post-TC has "EPSG:4326"?**

**ANSWER:**
The Pre-TC (Pre-Terrain Correction) image is still in the radar's native Range-Doppler geometry (slant range or ground range coordinates) rather than geographic map coordinates. Because it has not yet been geocoded or projected onto a terrestrial map, it lacks spatial orientation attributes, resulting in a CRS of "None". The post-TC (post-terrain correction) image has gone through the range-Doppler terrain correction process in SNAP. This process matches the radar coordinates to a digital elevation model (DEM) and projects the pixels onto a standard geographic coordinate reference system, which in this case is WGS 84 (EPSG:4326).

**QUESTION (10pts): Why can't we overlay the Pre-TC image with the DEM? Give two reasons.**

**ANSWER:**
We cannot overlay the Pre-TC image with the DEM for the following two reasons: geometry/coordinate mismatch: The Pre-TC image is in radar geometry (range-doppler coordinate space), where pixels are arranged by time-of-flight and Doppler shift. Conversely, the DEM is in a geographic map geometry (WGS 84) based on latitude and longitude. You cannot directly overlay two datasets that exist in completely different spatial domains.  Lack of Spatial Referencing (no CRS): As shown in the notebook outputs, the Pre-TC image has a CRS of None and lacks a proper geotransform matrix. Without geographic coordinates, map bounds, or a projection system attached to the image, a GIS or mapping library (like rasterio) cannot align it spatially over a georeferenced dataset like the DEM.

## Reflection question

**QUESTION (10pts): If a researcher downloaded a ready-made GRD product and never thought about terrain effects, what kinds of interpretation errors might they make in a mountainous study area?**

**ANSWER:**
Without terrain correction, a researcher will commit severe geometric and radiometric interpretation errors. Geometrically, because uncorrected data exists in radar-space, it will fail to align with standard GIS maps, land-use layers, or GPS data. Radar geometry also causes severe feature displacement; mountain peaks appear to lean toward the satellite (layover), shifting locations by hundreds of meters, while foreshortening compresses sensor-facing slopes and leads to a significant underestimation of their true surface area. Radiometrically, topography heavily distorts backscatter intensity, creating artificially bright radar-facing slopes and pitch-black radar shadows. A researcher unaware of these terrain effects might easily misinterpret these purely topographic artifacts as actual land-cover features, mistakenly identifying compressed, bright slopes as urban areas or shadow zones as water bodies.
